```python
%pip install networkx matplotlib

import networkx as nx
import matplotlib.pyplot as plt
from IPython.display import clear_output
import time

# Crear el grafo
G = nx.Graph()

edges = [
    ('a', 'b', 2),
    ('a', 'd', 2),
    ('a', 'f', 1),
    ('b', 'c', 2),
    ('b', 'd', 2),
    ('b', 'e', 4),
    ('c', 'e', 3),
    ('c', 'z', 1),
    ('d', 'e', 4),
    ('d', 'f', 3),
    ('e', 'g', 7),
    ('f', 'g', 5),
    ('g', 'z', 6)
]

G.add_weighted_edges_from(edges)

pos = {
    'a': (0, 1),
    'b': (1, 2),
    'c': (2, 2),
    'd': (1, 1),
    'e': (2, 1),
    'f': (1, 0),
    'g': (2, 0),
    'z': (3, 1)
}

# Algoritmo de Dijkstra paso a paso
def dijkstra_visual(G, start):
    # Inicializar etiquetas
    dist = {node: float('inf') for node in G.nodes}
    dist[start] = 0
    visited = set()
    steps = []

    while len(visited) < len(G.nodes):
        # Escoger el nodo no visitado con menor distancia
        current = min((node for node in G.nodes if node not in visited), key=lambda x: dist[x])
        visited.add(current)

        # Guardar el estado actual
        steps.append(dict(dist))  # copia del estado actual

        for neighbor in G.neighbors(current):
            if neighbor not in visited:
                weight = G[current][neighbor]['weight']
                if dist[current] + weight < dist[neighbor]:
                    dist[neighbor] = dist[current] + weight

    return steps

# Visualizar cada iteración
def draw_graph_with_labels(G, pos, steps, delay=1.5):
    for i, dist in enumerate(steps):
        clear_output(wait=True)
        plt.figure(figsize=(8, 5))
        nx.draw(G, pos, with_labels=True, node_color='skyblue', node_size=1500, font_size=14)
        edge_labels = nx.get_edge_attributes(G, 'weight')
        nx.draw_networkx_edge_labels(G, pos, edge_labels=edge_labels, font_size=12)
        
        # Mostrar etiquetas actuales (distancias)
        labels = {node: (f'{dist[node]}' if dist[node] != float('inf') else '∞') for node in G.nodes}
        nx.draw_networkx_labels(G, pos, labels=labels, font_size=14, font_color='black')

        plt.title(f"Iteración {i+1}", fontsize=16)
        plt.axis('off')
        plt.show()
        time.sleep(delay)

# Ejecutar Dijkstra y visualizar
steps = dijkstra_visual(G, 'a')
draw_graph_with_labels(G, pos, steps)
```
